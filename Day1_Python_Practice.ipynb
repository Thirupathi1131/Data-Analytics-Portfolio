```python
#Loops
# A loop is used to execute a block of code repeatedly or iteratively.

integers = [14,24,3,4,6]
mul2 = [] #empty list

# In Python, there are 2 main types of loops:
# 1.for loop = Used to iterate through a sequence like a list, string, tuple, or range.

for ele in integers :

  mul2.append(ele*2)

  #print(mul2) = it is inside the loop because it is indented.It executes 5 times, once for every element.

print(mul2) #This is outside the loop, so it executes only once, after all 5 iterations are finished.
```

```python
companies = {'google' :300, 'microsoft' :400, 'samsung' :250, 'apple' :450, 'openai' :1000}

#large cap, middle cap, small cap, large cap if > 350 crores

#get me a list of all the large cap companies [microsoft,apple,openai]

large_cap_companies=[]
for comp, price in companies.items():
  if price > 350:
    large_cap_companies.append(comp)

large_cap_companies
```
