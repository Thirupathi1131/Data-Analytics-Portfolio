```python
#IDE
#List

restaurants = ['kfc','paradise','abhiruchi','westin']

type(restaurants)

restaurants

restaurants.append('hosa') #The append() method is used to add a single element to the end of an existing list.

restaurants.extend(['mcd','true balck','kami']) #The extend() method is used to add multiple elements

len(restaurants) #returns the number of elements in a list

restaurants[1] #accessing an element from a list using index #Indexing starts from 0 (left to right).

restaurants[-1] #neagtive indexing #Negative indexing starts from -1 (right to left).

restaurants[:3]

restaurants[3:] #list slicing is used to access a specific portion of elements from a list.

restaurants[1:3]

restaurants.insert(0,'burger king') #Adds an element at a specified index

restaurants.remove('burger king') #Removes the first occurrence of a specified element from a list

restaurants.pop(1) #Removes and returns an element

restaurants.sort() #method arranges strings in alphabetical order (A–Z) by default.

restaurants.reverse()

restaurants

backup = restaurants.copy() #Create a backup

restaurants.clear() #restaurants.clear()
restaurants

restaurants = backup.copy() #Restore the list
restaurants

## restaurants.clear() removes all elements from a list, making it an empty list.
```

```python
#Tuple = A tuple is an ordered collection of elements that is immutable, meaning its elements cannot be changed after creation.

cars = ('BMW', 'Audi', 'Benz', 'Toyota')

type(cars)

cars

cars[0] #accessing elements

cars[-2]

cars[1:3]

cars.count('BMW')

cars.append('i20') #Tuples do not support append() because their elements cannot be changed after creation.
```
